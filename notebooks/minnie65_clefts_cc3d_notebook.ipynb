# Minnie65 clefts: 3000 x 3000 x 100 cuboid + cc3d count

This notebook pulls a 3000 x 3000 x 100 voxel cuboid from the Minnie65 clefts layer using CloudVolume,
thresholds the data to a binary cleft mask, and runs 3D connected components with cc3d to count
how many connected synapse objects are present in the cuboid.

The notebook uses cc3d's stack mode so the volume is processed slab-by-slab instead of loading the full
cuboid into RAM at once.

```python
from __future__ import annotations

from pathlib import Path
from typing import Iterator

import numpy as np
import cc3d
from cloudvolume import CloudVolume

try:
    import matplotlib.pyplot as plt
except Exception:
    plt = None
```
